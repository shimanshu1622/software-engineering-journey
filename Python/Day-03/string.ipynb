### What is String?
A string is a sequence of characters used to store text.

```python
name = "Python"
tea = 'masala tea'
```

### Type of Strings Quotes

```python
# Single Quotes
tea = 'Masala Tea'
# Double Quotes
tea = "Masala Tea"
# Triple Quotes
# Used for multi-line strings.
tea = """This
is
Python"""
```

### Indexing
Access characters using index numbers.

```python
tea = 'masala tea'
print(tea[0])
```

### String Slicing 
string[start:end]

```python
print(tea[2:4])
print(tea[-1])
print(tea[0:9:2])
```

```python
tea.lower() #lower(), upper()
```

```python
# strip() -> removes spaces from both ends

superhero = '      iron man    '
superhero.strip()
```
