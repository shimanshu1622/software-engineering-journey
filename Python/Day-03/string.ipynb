### What is String?
A string is a sequence of characters used to store text.

```python
name = "Python"
tea = 'masala tea'
```

### Type of Strings Quotes

```python
# Single Quotes
tea = 'Masala Tea'
# Double Quotes
tea = "Masala Tea"
# Triple Quotes
# Used for multi-line strings.
tea = """This
is
Python"""
```

### Indexing
Access characters using index numbers.

```python
tea = 'masala tea'
print(tea[0])
```

### String Slicing 
string[start:end]

```python
print(tea[2:4])
print(tea[-1])
print(tea[0:9:2])
```

```python
tea.lower() #lower(), upper()
```

```python
# strip() -> removes spaces from both ends

superhero = '      iron man    '
superhero = superhero.strip()
```

```python
# replace -> it used to replace the specific word from the string 
superhero.replace("iron","bat")
```

```python
# split() -> used to split the string/sentance in words based on space(default) or any other thing
chai = "Lemon Ginger black masala mint"
chai_list = chai.split() # ['Lemon', 'Ginger', 'black', 'masala', 'mint'] based on space
print(chai_list)
```

```python
chai_list2 = list(chai) # iterate through the string and each char is element of list even space 
print(chai_list2)
```

```python
chai_list3 = [ch for ch in chai] # same as above one 
print(chai_list3)
```

```python
# split custom delimeter
heros = "superman,ironman,batman,spiderman,thor,hulk" #split based on ,
heros_list = heros.split(",")
print(heros_list)
```

```python
# find() -> to find the specific word output will be index
print(heros.find("batman"))
```

```python
# count() -> tells the couont of word
print(heros.count("man"))
```

```python
# format -. used for formating 
chai_type = "masala"
quantity = 2
order = "I ordered {} cups of {} tea"
print(order.format(quantity,chai_type)) # I ordered 2 cups of masala tea
```

```python
# list to string 

print("".join(heros_list))
print(" ".join(heros_list))
print(", ".join(heros_list))
print("-".join(heros_list))
```

```python
# to use special char in strings use \ 
#chai = " he said, "masala chai is good" "
chai = " he said, \"masala chai is good\" "
print(chai)
```
