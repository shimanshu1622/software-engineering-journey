### Float point precision

````python
0.1 + 0.1 + 0.1 # 0.30000000000000004
````

````python
0.1 + 0.1 + 0.1 - 0.3 #5.551115123125783e-17
````

````python
# solve thios use decimal module
from decimal import Decimal
Decimal("0.1") + Decimal("0.1") + Decimal("0.1") 
Decimal("0.1") + Decimal("0.1") + Decimal("0.1") - Decimal("0.3")
````

Decimal should be initialized with strings bcz pythin floats are already approximated using binary floating-point representation.
passing a string allows decimal to create an exact decimal value without inheriting float-point precision error.



````python
print(Decimal(0.1))#0.1000000000000000055511151231257827021181583404541015625 inherits float error
print(Decimal("0.1")) #0.1
````

### Number Bases

````python
print(0b1010) #binary 0b
print(0o20) #octal 0o
print(0xff) #hexa 0x
````

````python
#convert
print(bin(15)) # binary 0b1111
print(oct(15)) # octal 0o17
print(hex(15)) # hexa 0xf
````
