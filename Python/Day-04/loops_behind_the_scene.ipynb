### what is iteration means
Iteration simply means repeating an operation
```
for i in range(5):
    print(i)
```
Iteration = looping over data one element at a time.

### Iteration tools in python
- for
- list comprehension
- map()
- filter()
- zip()
These tools can work on iterable obajects

### Iterable objects
An iterable is any object whose elements can be accessed one by one
- list 
- tuple
- string
- dict
- set
- range
- file object

### How does a ```for``` loop work internally?
```
my_list = [1,2,3]

for item in my_list:
    print(item)
```

- it does not directly access list elemets
**Internally:** Iterable object -> iter() -> Iterator -> next() -> Next element

````python
numbers = [10,20,30]
it = iter(numbers)
print(next(it))
print(next(it))
print(next(it))
# print(next(it)) # StopIteration Exeption error bcz there is no next item
````

### iter() func : 
- creates an iteraor from an iterable
- ```iterator = iter(iterable)```
- above ```it``` is an iterator

### next() func:
- returns the next value from an iterator

### Why for loop doesn't Crash?
bcz interanally
```
try:
    next(iterator)
except StopIteration:
    stop loop
```
- for loop handles this exception automatically
