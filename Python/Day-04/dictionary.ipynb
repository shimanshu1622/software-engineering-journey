```python
user1 = {"user":"shadow", "email":"shadow@mail.xyz", "age":22, "profile":"alpha"}
```

```python
# we can get value of the key via get method
print(user1.get("profile"))
```

```python
print(user1["user"])
```

```python
# adding new key:value and updating value of existing
user1["batch"] = 2026
print(user1)
```

```python
user1["age"] = 23
print(user1)
```

```python
#removing dict items
#1 del : removes item using key
del user1["age"]
print(user1)
```

```python
#2 pop() : removes item with the given key and return its value
val = user1.pop("batch")
print(user1)
print(val)
```

```python
#3 popitem() : removes and return the last inserted key-value pair
val1 = user1.popitem()
print(user1)
print(val1)
```

```python
#4 clear() : removes all items from dictionary
user1.clear()
print(user1)
```

```python
user1 = {"user":"shadow", "email":"shadow@mail.xyz", "age":22, "profile":"alpha"}
```

```python
# loops
for ele in user1:
    print(ele)

# as you notice key is printed but in list value was printed i.e. it go through keys
```

```python
# with help og get() we know we can access value 
for ele in user1:
    print(user1[ele])
```

```python
# for both key value pair 
for ele in user1:
    print(ele, user1[ele])
```

```python
for key, value in user1.items():
    print(key,value)
```

```python
# we can also check 
if "age" in user1:
    print("YES")
```
