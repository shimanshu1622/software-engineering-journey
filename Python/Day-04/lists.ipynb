```python
tea_varities = ["black", "green", "oolong", "white"]
print(tea_varities)
```

```python
# can access elements via indexing
print(tea_varities[1])
print(tea_varities[-1]) # last index
```

```python
# we can use slicing
print(tea_varities[1:3])
print(tea_varities[1:])
print(tea_varities[:2])
```

```python
# can modify value in list via indexing
tea_varities[1] = "herbal"
print(tea_varities)
```

```python
# what we happen if we do this
print(tea_varities[1:2])
tea_varities[1:2] = "lemon" #expected lemon at 1st index but ['black', 'l', 'e', 'm', 'o', 'n', 'oolong', 'white']
print(tea_varities)
# how to solve this 
tea_varities = ["black", "green", "oolong", "white"]
tea_varities[1:2] = ["lemon"] # pass it in list[]
print(tea_varities)
```

```python
print(tea_varities[1:3])
tea_varities[1:3] = ["masala"]
print(tea_varities)
```

```python
# we can insert values via slicing also
tea_varities[1:1] = ["green"]
print(tea_varities)
```

```python
# we can remove also via slicing by inserting nothing
tea_varities[2:3] = [] # remove the 2nd index or say replace it by nothing
print(tea_varities)
```

```python
for t in tea_varities:
    print(t)
# it will print all the elements in seprate lines one by one
```

```python
# if i want it in same line 
for tea in tea_varities:
    print(tea,end=" ") # this will give output in one line with space as i mentioned we can put anything there
```

```python
# append -> add the element at the last
tea_varities.append("oolong")
print(tea_varities)
```

```python
# pop() used to remove the last element
tea_varities.pop()
print(tea_varities)
```

```python
# remove is used to remove the specific element 
tea_varities.remove("green")
print(tea_varities)
# but you run it twice for the same element it will give the error 
```

```python
# insert() use to add element at specific index
tea_varities.insert(1,"green")
print(tea_varities)
```

```python
# tea_copy = tea_varities  # both have same reference change made in one show in both
tea_copy = tea_varities.copy() # both have different reference
```
