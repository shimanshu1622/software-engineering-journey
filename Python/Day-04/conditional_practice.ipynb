### Age Group Categorization
Classify a person's age group:
- child ( < 13 )
- Teenager ( 13 - 19 )
- Adult ( 20 - 59 )
- Senior (60+)

```python

def age_category():
    age = int(input("provide the Age"))

    if age < 13:
        print("Child")
    elif age < 20:
        print("teenager")
    elif age < 59:
        print("adult")
    else:
        print("senior")

age_category() # 23 -> adult
```

### Movie Ticket Pricing
movie ticket are priced on age: 
- $12 for adults (18 and above)
- $8 for children
- $2 discount on Wednesday Everyone

```python
def movie_ticket():
    day = input("Enter Day of Booking").strip().lower()
    number_of_ticket = int(input("Enter the number of tickets to book"))
    price = 0
    for num in range(number_of_ticket):
        age = int(input(f"Enter age of person {num+1}"))
        if age >= 18: price += 12
        else: price += 8
    if day == "wednesday": 
        discount = 2*number_of_ticket
        price -= discount

    print(f"Total Payment: {price}")

movie_ticket()

```
