### Age Group Categorization
Classify a person's age group:
- child ( < 13 )
- Teenager ( 13 - 19 )
- Adult ( 20 - 59 )
- Senior (60+)

```python

def age_category():
    age = int(input("provide the Age"))

    if age < 13:
        print("Child")
    elif age < 20:
        print("teenager")
    elif age < 59:
        print("adult")
    else:
        print("senior")


age_category() # 23 -> adult
```

### Movie Ticket Pricing
movie ticket are priced on age: 
- $12 for adults (18 and above)
- $8 for children
- $2 discount on Wednesday Everyone

```python
def movie_ticket():
    day = input("Enter Day of Booking").strip().lower()
    number_of_ticket = int(input("Enter the number of tickets to book"))
    price = 0
    for num in range(number_of_ticket):
        age = int(input(f"Enter age of person {num+1}"))
        if age >= 18: price += 12
        else: price += 8
    if day == "wednesday": 
        discount = 2*number_of_ticket
        price -= discount

    print(f"Total Payment: ${price}")

movie_ticket()

```

### Fruit Ripeness Checker
Determine if a fruit is ripe, overripe or unripe based on its color(e.g Banana: Green - unripe, yellow - ripe, brown - overripe)

```python
def fruit_ripeness():
    fruit = input("Enter the name of fruit: ").strip().lower()
    color = input("what is the color of fruit: ").strip().lower()
    if fruit=="banana":
        if color=="green" : print("Unripe")
        elif color=="yellow" : print("Ripe")
        elif color=="brown": print("Overripe")

fruit_ripeness()
```

```python
# sum of even numbers upto given n

def even_sum():
    num = int(input("enter the number"))
    sum = 0
    for i in range(1,num+1):
        if i % 2 ==0 : sum += i
    print(f"Even sum for numbers upto {num} is: {sum}")

even_sum()
```

```python
# the multiplication table for given number upto 10 , but skip the 5th iteration
def table_print():
    num = int(input("Enter the Number"))
    for i in range(1,11):
        if i != 5 : print(f"{num} X {i} = {num*i}")

table_print()
    
```

```python
# reverse the string using loop
def reverse_string():
    str = input("Enter the string")
    reversed_str = ""

    for char in str:
        reversed_str = char + reversed_str

    print(reversed_str)


reverse_string()
```

```python
# find the first non -repiting character
def first_non_repiting_char1():
    str = input("Enter the string")

    for char in str:
        if str.count(char) == 1 :
            print(f"Char is : {char}")
            break

first_non_repiting_char1()

```

```python
# 2ND method 

def first_non_repiting_char_by_dict():
    str = input("enter the string")
    dict = {}

    for char in str:
        dict[char] = dict.get(char,0) +1

    for char in str:
        if dict[char] == 1:
            print(f"Char is : {char}")
            break

first_non_repiting_char_by_dict()
```

```python
# factorial of number

def factorial():
    n = int(input("enter the number: "))
    result = 1
    while n > 0:
        result = result * n
        n -= 1
    print(result)


factorial()
```

```python
# prime number 

def prime_number():
    num = int(input("Enter the number"))
    not_prime = False
    for i in range(2 , num): # for i in range(2, int(num ** 0.5) + 1)
        if num % i == 0 : 
            not_prime = True
            break
    if not_prime : print(f"{num} is NOT a PRIME number")
    else:  print(f"{num} is PRIME number")

prime_number()
```

```python
# find if the list is unique or NOT
def unique_item():
    items = ["apple" , "banana", "orange", "apple", "mango"]

    unique = set()

    for item in items:
        if item in unique:
            print("Duplicate found: ",item)
            break
        unique.add(item)

unique_item()
```
