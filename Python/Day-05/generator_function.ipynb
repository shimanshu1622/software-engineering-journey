### What is Generator Function?
- a agenrator function is a function that produces values one by one instead of returning everything at once.
#### ```yield``` is a keyword that:
- returns the current value
- pauses the function
- saves the function's current state in memory
- continues from the same point when called again


````python
def even_num_generator(limit):
    for i in range(2,limit+1,2):
        yield i
````

````python
even_num_generator(2)
````

````python
for num in even_num_generator(10):
    print(num)
````

### Why not use return?


````python
def even_num_generator2(limit):
    for i in range(2,limit+1,2):
        return i
````

````python
even_num_generator2(10)
````

2 -> only the first value is returned
**why?**
- return -> immediately terminates the function

```without yield```
- creates List -> store all values -> return complete value
- everything is stored before returning.
```with yield```
- generate 2 -> use it -> generate 4 -> use it -> generate 6 -> use it -> so on.....
- only one value exits at a time
- much less memory is used

### Generator object
- calling a genartor function does not immediatly execute the function
- instead, python creates a generator object
- values are produced only when needed.

### When should you use generators?
- processing large files 
- reading huge datasets
- working with streams of data
- producing values sequentially
- saving memory
### Avoid generators
- you need random access to all values
- you need the complete collection immediately
