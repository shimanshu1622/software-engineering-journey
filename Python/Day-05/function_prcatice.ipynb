### What is function?
A function is a reusable block of code that performs a specific task. <br>
Instead of writing the same code multiple times, we write it once inside a function and call it whenever needed.

````python
def square(num):
    return num ** 2

print(square(4))
````

````python
# print() vs return
def double(num):
    print("double is :",num *2)

print(double(3)) # function return nothing thats why None is there and print is in the function body
````

````python
def add(num1,num2):
    return num1+num2

print(add(4,5))
````

### parameter vs argumnent
- num1 , num2 is a parameter 
- 4,5 is argument


````python
# local variables -. variables created inside the function body exits only duirng the function's execution.
def square(num):
    result = num ** 2
    return result  # here result is local varible outside the function reuslt is not present
````

````python
# can return multiple values 
import math
def cicle_stats(radius):
    area = math.pi * radius * radius
    circumference = 2 * math.pi * radius
    # return area,circumference
    return round(area,2), round(circumference,2)

a,c = cicle_stats(3)
print(a,c)
````

````python
# default parameters 
def greet(name = "Guest"):
    return f"Hello, {name}!"

print(greet())
print(greet("Shadow"))
````

````python
# multiple default parameters
def student(name, course="python",duration="15days"): # required parameter comes first always 
    return name,course,duration

student("Shadow")
````

````python
student("Super","Backend","3-month")
````

````python
student(duration="6-month",name="Chad",course="System design")
````

### Lambda Function (Anonymous Function)
A Lambda Function is a small anonymous function (a function without a name).<br>
It is generally used when you need a function only once or for a short operation.

````python
def cube(x):
    return x**3
#lambda parameters:expression
cube2 = lambda x:x**3
````

````python
# multiple parameters
add = lambda a,b : a+b
print(add(4,7))
````

#### Where lambda functions is used?
- django
- Flask
- Python libraries
- Frameworks
- Functional programming utilities

#### When should use lambda?
- function is very small
- used only once
- a one - line calculation is enough
#### Avoid
- logic is large
- multiple statements are required
- readability becomes poor


### Variable-Length Arguments (*args)

````python
def add(a,b):
    return a+b
add(3,6)
# add(3,4,5,6) # does not work 
# bcz function accepts only 2 parameters 
````

Suppose you want a function that can add: n numbers 
- So use ***args** 
- ***args** allows a function to accept any number of positional argumets.

````python
def summ_all(*args):
    return sum(args)
````

````python
print(summ_all(2,3,4))
print(summ_all(2,3,4,6,3,56))
print(summ_all(2,3,4,1,54,56,87))
print(summ_all(2,3))
````

### Why the ```*```
The ```*``` is what gives the specila behavior, not the name ```args```.

````python
def demo(*args):
    return args

def demo2(*chai):
    return chai

# technically both are valid 
# BUT, Industry practice is using args
````

### What is ```args``` Internally
- ```args``` is actually stored as a tuple.

````python
print(demo(1,2,3)) # (1,2,3)
````

- Notice the parantheses
- that means ```args``` is a **tuple**
- Since ```args``` is tuple we can loop over it.

````python
def multiple_by_two(*args):
    for i in args:
        print(i*2)

multiple_by_two(1,2,'A')
````

### Keyword varibale-Length Arguments (**kwargs)

````python
# *args allows us to pass multiple positional arguments
def demo(*args):
    print(args)

demo(1,2,3,4)
````

- BUT what if we want to pass information like
```
name = "shadow"
age = 21
city = "Noida"
```
- the number of these values may vary

### What is ```**kwargs```
- ```**kwargs``` allows a function to accepts any number of keywords arguments.

````python
def student(**kwargs):
    print(kwargs)

student(
    name = "shadow",
    age = 21,
    city = "Noida"
)
````

### Why the ```**```
- just like ```*args```, the special behavior comes from the ```**```
- the name ```kwargs``` is only a convention.

### What is ```kwargs``` Internally
- Python stored keyword argumets inside a dictionary.

````python
def demo(**kwargs):
    print(type(kwargs))

demo() # <class 'dict'>
````

````python
def students(**kwargs):
    for key,value in kwargs.items():
        print(key,":",value)
````

````python
students(name='shadow', age= 21, city= 'Noida')
````
